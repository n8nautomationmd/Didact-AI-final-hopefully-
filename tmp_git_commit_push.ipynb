```python
import subprocess
from pathlib import Path
repo = Path('/workspaces/Didact-AI-final-hopefully-')
print('PWD:', repo)
print('Git status:')
print(subprocess.run(['git', '-C', str(repo), 'status', '--short'], capture_output=True, text=True).stdout)
print('Git diff app.py:')
print(subprocess.run(['git', '-C', str(repo), 'diff', '--', 'app.py'], capture_output=True, text=True).stdout)
print('Committing changes...')
subprocess.run(['git', '-C', str(repo), 'add', 'app.py'], check=True)
subprocess.run(['git', '-C', str(repo), 'commit', '-m', 'Remove Echipă & Etică AI tab from app.py'], check=True)
print('Pushing to remote...')
print(subprocess.run(['git', '-C', str(repo), 'push'], capture_output=True, text=True).stdout)
```
